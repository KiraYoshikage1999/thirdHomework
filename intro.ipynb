```python
import matplotlib.pyplot as plt
import numpy as np
```

##Task1 

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plan_sales = [100, 120, 130, 150, 170, 160, 180, 190, 200, 210, 220, 250]
fact_sales = [95, 125, 120, 155, 165, 175, 170, 195, 210, 205, 230, 260]

plt.figure(figsize=(10, 5))
plt.plot(months, plan_sales, label='Plan', marker='o', color='blue', linestyle='--')
plt.plot(months, fact_sales, label='Fact', marker='s', color='green', linestyle='-')

plt.title('Comparison of Planned and Actual Sales by Month')
plt.xlabel('Month')
plt.ylabel('Sales Volume')
plt.legend()
plt.grid(True)
plt.show()
```

##Task 2

```python
np.random.seed(42)
ages = np.random.randint(18, 71, size=100)

mean_age = np.mean(ages)

plt.figure(figsize=(9, 5))
plt.hist(ages, bins=10, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(mean_age, color='red', linestyle='dashed', linewidth=2, label=f'Mean Age: {mean_age:.1f}')

plt.title('Age Distribution of 100 People')
plt.xlabel('Age')
plt.ylabel('Number of People')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()
```

##Task 3 

```python
np.random.seed(0)
group_a = np.random.normal(75, 10, 30)
group_b = np.random.normal(82, 8, 30)
group_c = np.random.normal(68, 15, 30)

data_groups = [group_a, group_b, group_c]
group_labels = ['Group A', 'Group B', 'Group C']

plt.figure(figsize=(8, 5))
plt.boxplot(data_groups, tick_labels=group_labels, patch_artist=True,
            boxprops=dict(facecolor='lightgreen', color='green'),
            medianprops=dict(color='red', linewidth=2))

plt.title('Comparison of Score Spread Between Student Groups')
plt.xlabel('Groups')
plt.ylabel('Scores')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

##Task 4

```python
print("=== Task 4 ===")
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
temperature = [22, 24, 19, 21, 25, 27, 26]
humidity = [65, 60, 75, 70, 55, 50, 53]

fig, ax1 = plt.subplots(figsize=(10, 5))

color = 'tab:red'
ax1.set_xlabel('Days of the Week')
ax1.set_ylabel('Temperature (°C)', color=color)
line1 = ax1.plot(days, temperature, color=color, marker='o', label='Temperature', linewidth=2)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Humidity (%)', color=color)
line2 = ax2.plot(days, humidity, color=color, marker='s', linestyle='--', label='Humidity', linewidth=2)
ax2.tick_params(axis='y', labelcolor=color)

# Combine legends from both axes
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Daily Temperature and Humidity Over a Week')
fig.autofmt_xdate() 
plt.grid(True, alpha=0.3)
plt.show()
```

##Task 5

```python
hours = list(range(24))
server_load = [15, 10, 8, 7, 6, 9, 20, 45, 70, 85, 90, 88, 
               85, 82, 80, 83, 92, 95, 90, 75, 60, 45, 30, 20]

plt.figure(figsize=(10, 5))
plt.plot(hours, server_load, color='purple', linewidth=2, label='Load')
plt.fill_between(hours, server_load, color='purple', alpha=0.3)

plt.title('Hourly Server Load Over 24 Hours')
plt.xlabel('Hours (0-23)')
plt.ylabel('Load (%)')
plt.xticks(hours)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

##Task 6

```python
time_scale = ['Week 1', 'Week 2', 'Week 3', 'Week 4']
conversion = [2.1, 2.3, 2.5, 2.8]
retention = [40, 42, 41, 45]
avg_check = [500, 520, 510, 540]
orders = [1200, 1350, 1300, 1500]

fig, axs = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Product Metrics Over Time', fontsize=16)

# 1. Conversion
axs[0, 0].plot(time_scale, conversion, marker='o', color='teal')
axs[0, 0].set_title('Conversion (%)')
axs[0, 0].grid(True)

# 2. Retention
axs[0, 1].plot(time_scale, retention, marker='s', color='darkorange')
axs[0, 1].set_title('Retention (%)')
axs[0, 1].grid(True)

# 3. Average Check
axs[1, 0].plot(time_scale, avg_check, marker='^', color='crimson')
axs[1, 0].set_title('Average Check')
axs[1, 0].grid(True)

# 4. Number of Orders
axs[1, 1].plot(time_scale, orders, marker='d', color='dodgerblue')
axs[1, 1].set_title('Number of Orders')
axs[1, 1].grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
